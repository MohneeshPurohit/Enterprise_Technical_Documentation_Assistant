# Capstone Notebook 03: RAG Pipeline Execution & Quantitative Benchmark Evaluation
**Objective**: Execute full end-to-end RAG queries, test Grounded Prompting, verify Multi-Turn Memory, demonstrate Hallucination Guardrails, and calculate quantitative evaluation metrics.

```python
import sys, os
sys.path.append(os.path.abspath('..'))

from src.rag_pipeline import TechnicalRAGPipeline

# Initialize RAG Pipeline
pipeline = TechnicalRAGPipeline('../vectorstore')

# Execute In-Domain Technical Query
user_query = "What causes HTTP 401 Unauthorized error and how do I fix it?"
response = pipeline.run(user_query)

print("=== AI GROUNDED RESPONSE ===")
print(response['answer'])
```

```python
# Test Multi-Turn Conversational Memory & Query Reformulation
chat_history = [
    {"role": "user", "content": "How do I setup webhooks?"},
    {"role": "assistant", "content": "Use HMAC-SHA256 signatures passed in the header."}
]

followup_query = "How do I verify the signature?"
print(f"Follow-Up Question: '{followup_query}'\n")

response_memory = pipeline.run(followup_query, chat_history=chat_history)
print("=== MEMORY-REFORMULATED RESPONSE ===")
print(response_memory['answer'])
```

```python
# Test Hallucination Refusal Guardrail (Out-of-Domain Query)
out_of_domain_query = "What is the employee annual vacation leave policy?"
response_guardrail = pipeline.run(out_of_domain_query)

print(f"Out-of-Domain Query: '{out_of_domain_query}'\n")
print("=== GUARDRAIL REFUSAL OUTPUT ===")
print(response_guardrail['answer'])
```

```python
# Execute Quantitative RAG Evaluation Benchmark Suite
from src.evaluation import RAGEvaluator

evaluator = RAGEvaluator(pipeline)
summary = evaluator.evaluate()

print("\n================ FINAL CAPSTONE EVALUATION METRICS ================")
print(f"Total Benchmark Tests     : {summary['total_benchmark_tests']}")
print(f"Retrieval Hit Rate @ K    : {summary['retrieval_hit_rate']}%")
print(f"Answer Groundedness %     : {summary['answer_accuracy_rate']}%")
print(f"Hallucination Refusal %   : {summary['hallucination_refusal_rate']}%")
print("===================================================================")
```
