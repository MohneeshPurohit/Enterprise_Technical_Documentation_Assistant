# Capstone Notebook 01: Document Processing & Exploration Pipeline
**Domain**: Enterprise Technical & API Documentation Intelligence  
**Objective**: Parse raw technical PDF documents, strip formatting artifacts, preserve technical code syntax, and extract detailed page-level metadata.

```python
import sys, os
sys.path.append(os.path.abspath('..'))
from src.document_loader import DocumentLoader

loader = DocumentLoader('../data/documents')
documents = loader.load_all_documents()
print(f"Total Ingested Pages/Sections: {len(documents)}")
```

```python
# Inspect Page-Level Metadata & Text Statistics
for idx, doc in enumerate(documents, start=1):
    meta = doc['metadata']
    print(f"=== Document Page {idx} ===")
    print(f"File Name   : {meta['source']}")
    print(f"Page Number : {meta['page']} of {meta['total_pages']}")
    print(f"Char Count  : {meta['char_count']} characters")
    print(f"File Type   : {meta['file_type']}")
    print("-" * 50)
    print("Cleaned Text Preview:\n" + doc['content'][:300] + "\n" + "=" * 50)
```
